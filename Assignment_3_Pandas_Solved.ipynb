## Q6. Create employees.csv

```python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
})

employees.to_csv("employees.csv", index=False)
employees
```

### d) First 5 rows and last 3 rows

```python
print("First 5 rows:")
display(employees.head())

print("Last 3 rows:")
display(employees.tail(3))
```

### e) Statistics

```python
average_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("Average salary:", average_salary)
print("Total bonus:", total_bonus)
print("Youngest employee's age:", youngest_age)
print("Highest performance rating:", highest_rating)
```

### f) Sort by Salary in descending order

```python
employees_sorted = employees.sort_values("Salary", ascending=False)
employees_sorted
```

### g) Categorize employees based on performance rating

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    return "Average"

employees["Performance"] = employees["Rating"].apply(performance_category)
employees
```

### h) Identify missing values

```python
employees.isnull().sum()
```

### i) Rename Employee_ID to ID

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
employees
```

### j) Find employees with more than 5 years of experience and employees from IT

```python
more_than_5_years = employees[employees["Years_of_Experience"] > 5]
it_employees = employees[employees["Department"] == "IT"]

print("Employees with more than 5 years of experience:")
display(more_than_5_years)

print("Employees in the IT department:")
display(it_employees)
```

### k) Add Tax column

```python
employees["Tax"] = employees["Salary"] * 0.10
employees
```

### l) Save the modified DataFrame

```python
employees.to_csv("employees_modified.csv", index=False)
employees
```
