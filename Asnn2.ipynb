```python

roll_no = input("Enter your roll number: ")

L = [int(digit) * 10 for digit in roll_no]

print("Initial L:", L)

L.append(50)
print("After append(50):", L)

L.insert(2, 70)
print("After insert(2, 70):", L)

L.remove(50)
print("After remove(50):", L)

removed = L.pop(2)
print("After pop(2):", L)

L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)



print("First three:", L[:3])
print("Last three:", L[-3:])
 

average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)

scores = tuple(L[:8])

print("\nScores tuple:", scores)


```

```python
# i. Highest score and its index
highest = max(scores)
highest_index = scores.index(highest)

# Lowest score and number of occurrences
lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Highest score index:", highest_index)

print("Lowest score:", lowest)
print("Lowest score occurrences:", lowest_count)


# ii. Reverse tuple and return as a list

# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(scores[::-1])

print("Reversed tuple as list:", reversed_scores)


# iii. Ask user for a score

user_score = int(input("\nEnter a score to search: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")


# iv. Attempt to change tuple element

try:
    scores[0] = 100
except TypeError as e:
    print("\nError:", e)
    # Tuples are immutable, so their elements cannot be changed directly;
    # lists are mutable and allow element assignment.


# v. Tuple unpacking using *

first_score, second_score, *remaining_scores = scores

print("\nFirst score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# =========================
# Q3 - RANDOM NUMBERS
# =========================

import random

# Enter your roll number
roll_no = int(input("Enter your roll number: "))

# Set random seed
random.seed(roll_no)

# i & ii. Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)


# iii. Count and print all odd numbers

odd_numbers = [x for x in numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))


# iv. Count and print all even numbers

even_numbers = [x for x in numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))


# v. Prime numbers

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


# Use list comprehension to create the prime list
prime_numbers = [x for x in numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))


# vi. Number occurring most frequently

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequent number:", most_frequent)
print("Number of occurrences:", frequency)
```

```python
# =========================
# Q4 - SET OPERATIONS
# =========================

# Enter your roll number
roll_no = input("Enter your roll number: ")

# Extract the 8 digits
digits = [int(digit) for digit in roll_no[:8]]

print("Original digits:", digits)


# v. Create sets A and B

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("\nSet A:", A)
print("Set B:", B)


# vi. Union of A and B
# Contains all unique values from both sets

union = A.union(B)

print("\nUnion:", union)


# vii. Intersection of A and B
# Contains only values common to both sets

intersection = A.intersection(B)

print("Intersection:", intersection)


# viii. Difference

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("\nA - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present in one set but not the other
# for a specific direction, while symmetric_difference() combines
# both differences.


# ix. Symmetric difference

symmetric = A.symmetric_difference(B)

print("\nSymmetric difference:", symmetric)


# x. Subset and superset

print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))


# xi. Ask user for X and discard it from A

X = int(input("\nEnter a value X to remove from A: "))

A.discard(X)

print("A after discard:", A)

# discard() is safer because it does nothing if X is not present,
# whereas remove() raises a KeyError.
```

```python
# =========================
# Q5 - DICTIONARY OPERATIONS
# =========================

# Original dictionary
my_dict = {
    "name": "Your Name",
    "roll_no": "Your Roll Number",
    "branch": "CSE",
    "age": 20,
    "city": "Your Home City"
}

print("Original dictionary:")
print(my_dict)


# i. Rename "city" to "location" using pop()

my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)


# ii. Add CGPA

my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)


# iii. Increase age by 1

my_dict["age"] += 1

print("\nAfter increasing age by 1:")
print(my_dict)


# iv. Delete "branch" using pop() and del
# Make two separate copies

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Using pop()
removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Returned value:", removed_branch)

# Using del
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.


# v. Iterate using .items()

print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)


# vi. Check whether "email" exists

if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")


# vii. Create friend's dictionary

friend_dict = {
    "name": "Rahul",
    "roll_no": "CS102",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}

# Merge using ** syntax
merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# If both dictionaries have the same key,
# the value from the second dictionary (friend_dict) wins.


# viii. Dictionary comprehension
# Keep only key-value pairs whose values are strings

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)
```
