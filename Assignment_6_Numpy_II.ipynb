```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print(corrected_temperature)

fahrenheit = (9 / 5) * corrected_temperature + 32
print(fahrenheit)

readings_above_32 = corrected_temperature[corrected_temperature > 32]
print(readings_above_32)

count_above_32 = np.sum(corrected_temperature > 32)
print(count_above_32)

print("Vectorization applies operations to an entire array at once, reducing Python-level loop overhead. Boolean indexing selects values that satisfy a condition directly, making filtering concise and efficient.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(np.sum(steps))
print(np.mean(steps))
print(np.max(steps))
print(np.min(steps))
print(np.sum(steps, axis=0))
print(np.sum(steps, axis=1))
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print(max_position)
```

```python
original = np.array([1, 2, 3, 4, 5, 6])
print(original)
```

```python
subset = original[1:5]
print(subset)
```

```python
subset[0] = 999
print(original)
print(subset)
```

```python
copied_subset = original[1:5].copy()
copied_subset[0] = 500
print(original)
print(copied_subset)
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
```

```python
print(matrix[0])
```

```python
print(matrix[-1])
```

```python
print(matrix[:, 1])
```

```python
print(matrix[1:3, 2:4])
```

```python
flattened = matrix.flatten()
raveled = matrix.ravel()
print(flattened)
print(raveled)
```

```python
raveled[0] = 100
print(matrix)
print(raveled)
```

```python
flattened[1] = 200
print(matrix)
print(flattened)
```
